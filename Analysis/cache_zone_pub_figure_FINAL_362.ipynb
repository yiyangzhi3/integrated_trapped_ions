```python
def mark_waist(ax, fit_params, coordinate_offset, color="C0"):
    center = fit_params["center_um"] - coordinate_offset
    waist = fit_params["waist_um"]

    # Intensity at one waist radius from the fitted center.
    level = (
        fit_params["background"]
        + fit_params["amplitude"] * np.exp(-2)
    )

    ax.annotate(
        "",
        xy=(center + waist, level),
        xytext=(center, level),
        arrowprops=dict(
            arrowstyle="->",
            color=color,
            linewidth=1.5,
            shrinkA=0,
            shrinkB=0,
        ),
    )

    ax.annotate(
        f"{waist:.2f} µm",
        xy=(center + waist / 2, level),
        xytext=(-12, 10),
        textcoords="offset points",
        ha="center",
        color=color,
        fontsize=11,
    )
```

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit


# ------------------------------------------------------------
# Gaussian intensity fit:
# I(s) = background + amplitude * exp(-2 * ((s - center) / waist)**2)
# ------------------------------------------------------------

def gaussian(position, amplitude, center, waist, background):
    return background + amplitude * np.exp(
        -2 * ((position - center) / waist) ** 2
    )


def fit_camera_slice(position, intensity):
    position = np.asarray(position, dtype=float)
    intensity = np.asarray(intensity, dtype=float)

    amplitude = np.ptp(intensity)
    background = intensity.min()
    center = position[np.argmax(intensity)]

    if amplitude <= 0:
        raise ValueError("A constant profile has no measurable Gaussian waist.")

    # Initial waist estimate from the full width at half maximum.
    half_max_positions = position[
        intensity >= background + amplitude / 2
    ]
    waist = np.ptp(half_max_positions) / np.sqrt(2 * np.log(2))

    if waist == 0:
        waist = np.ptp(position) / 10

    parameters, _ = curve_fit(
        gaussian,
        position,
        intensity,
        p0=[amplitude, center, waist, background],
        bounds=(
            [0, position.min(), np.finfo(float).eps, -np.inf],
            [np.inf, position.max(), np.inf, np.inf],
        ),
        maxfev=10000,
    )

    amplitude, center, waist, background = parameters

    return {
        "amplitude": amplitude,
        "center_um": center,
        "waist_um": waist,
        "background": background,
        "curve": gaussian(position, *parameters),
    }


# ------------------------------------------------------------
# Figure settings
# ------------------------------------------------------------

N_ROWS = 3
N_COLS = 3
PANEL_WIDTH = 5
PANEL_HEIGHT = 5

window = 15

fig, axes = plt.subplots(
    N_ROWS,
    N_COLS,
    figsize=(N_COLS * PANEL_WIDTH, N_ROWS * PANEL_HEIGHT),
    dpi=100,
    constrained_layout=False,
    squeeze=False,
)

# Reserve space on the right for the profile legends.
fig.subplots_adjust(
    left=0.08,
    right=0.80,
    bottom=0.08,
    top=0.94,
    wspace=0.25,
    hspace=0.2,
)

datasets = [
    (Sim_TE_455, TE_450, cmap_450, "455 nm", "450 nm"),
    (Sim_TE_532, TE_515, cmap_515, "532 nm", "515 nm"),
    (Sim_TE_650, TE_635, cmap_635, "650 nm", "635 nm"),
]

# Camera fit results remain available after plotting.
camera_fits = {}


# ------------------------------------------------------------
# Plot each wavelength pair
# ------------------------------------------------------------

for row, (simulation, camera, cmap, design_wl, camera_wl) in enumerate(datasets):
    # Fit the current simulation slices.
    simulation.fit_gaussian_slices()

    # Fit the current camera slices.
    # Fits use the full slices; window only controls the displayed range.
    axial_fit = fit_camera_slice(
        camera.x_um,
        camera.axial_slice,
    )
    radial_fit = fit_camera_slice(
        camera.y_um,
        camera.radial_slice,
    )

    camera_fits[camera_wl] = {
        "axial": axial_fit,
        "radial": radial_fit,
    }

    # --------------------------------------------------------
    # Column 1: design image
    # --------------------------------------------------------

    ax = axes[row, 0]

    im = ax.pcolormesh(
        simulation.x_um - simulation.x0,
        simulation.y_um - simulation.y0,
        simulation.norm_P,
        shading="nearest",
        cmap=cmap,
        vmin=0,
        vmax=1,
    )

    fig.colorbar(im, ax=ax, shrink=0.8, pad=0.03)

    ax.set_aspect("equal")
    ax.set(
        xlim=(-window, window),
        ylim=(-window, window),
        title=f"TE Design ({design_wl})",
    )

    # --------------------------------------------------------
    # Column 2: measured image
    # --------------------------------------------------------

    ax = axes[row, 1]

    im = ax.pcolormesh(
        camera.x_um - camera.x0,
        camera.y_um - camera.y0,
        camera.height_slice,
        shading="nearest",
        cmap=cmap,
        vmin=0,
        vmax=1,
    )

    fig.colorbar(im, ax=ax, shrink=0.8, pad=0.03)

    ax.set_aspect("equal")
    ax.set(
        xlim=(-window, window),
        ylim=(-window, window),
        title=f"TE Measured ({camera_wl})",
    )

    # --------------------------------------------------------
    # Column 3: axial and radial profiles
    # --------------------------------------------------------

    slot = axes[row, 2].get_subplotspec()
    axes[row, 2].remove()

    inner = slot.subgridspec(2, 1, hspace=0.6)
    ax_top = fig.add_subplot(inner[0, 0])
    ax_bottom = fig.add_subplot(inner[1, 0])

    profiles = [
        (
            ax_top,
            simulation.x_um - simulation.x0,
            simulation.axial_slice,
            simulation.axial_waist_um,
            camera.x_um - camera.x0,
            camera.axial_slice,
            axial_fit,
            r"Axial position ($\mu$m)",
        ),
        (
            ax_bottom,
            simulation.y_um - simulation.y0,
            simulation.radial_slice,
            simulation.radial_waist_um,
            camera.y_um - camera.y0,
            camera.radial_slice,
            radial_fit,
            r"Radial position ($\mu$m)",
        ),
    ]

    for (
        ax,
        design_position,
        design_slice,
        design_waist,
        camera_position,
        camera_slice,
        camera_fit,
        xlabel,
    ) in profiles:
        ax.plot(
            design_position,
            design_slice,
            color="C0",
            linewidth=2,

            label=f"Design: $w$ = {design_waist:.2f} $\mu$m",
        )

        ax.plot(
            camera_position,
            camera_slice,
            color="C1",
            alpha=0.65,
            linewidth=2,

            label=f"Camera: $w$ = {camera_fit['waist_um']:.2f} $\mu$m",
        )

        ax.plot(
            camera_position,
            camera_fit["curve"],
            color="black",
            linestyle="--",
            linewidth=2,
            label="Camera Gaussian fit",
        )

        ax.set(
            xlim=(-window, window),
            xlabel=xlabel,
        )

        ax.legend(
            loc="center left",
            bbox_to_anchor=(1.02, 0.5),
            borderaxespad=0,
            frameon=False,
            fontsize=12,
        )


# ------------------------------------------------------------
# Shared labels for the first two columns
# ------------------------------------------------------------

for ax in axes[:, :2].flat:
    ax.set_xlabel("")
    ax.set_ylabel("")

fig.canvas.draw()

positions = [ax.get_position() for ax in axes[:, :2].flat]

left = min(p.x0 for p in positions)
right = max(p.x1 for p in positions)
bottom = min(p.y0 for p in positions)
top = max(p.y1 for p in positions)

label_style = {
    "fontsize": plt.rcParams["axes.labelsize"],
    "fontweight": plt.rcParams["axes.labelweight"],
    "color": plt.rcParams["axes.labelcolor"],
    "fontfamily": plt.rcParams["font.family"],
}

fig.text(
    (left + right) / 2,
    bottom - 0.035,
    r"Axial position ($\mu$m)",
    ha="center",
    va="top",
    **label_style,
)

fig.text(
    left - 0.045,
    (bottom + top) / 2,
    r"Radial position ($\mu$m)",
    rotation=90,
    ha="center",
    va="center",
    **label_style,
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar

plt.rcParams["svg.fonttype"] = "none"

def align_ion_to_camera(ion_x, intensity, camera, max_offset_um=20):
    ion_x = np.asarray(ion_x, dtype=float)
    intensity = np.asarray(intensity, dtype=float)

    camera_x = np.asarray(camera.x_um, dtype=float)
    camera_intensity = np.asarray(camera.axial_slice, dtype=float)

    # Start with both measured maxima at zero.
    ion_x = ion_x - ion_x[np.argmax(intensity)]
    camera_x = camera_x - camera_x[np.argmax(camera_intensity)]
    camera_intensity = camera_intensity / camera_intensity.max()

    order = np.argsort(camera_x)
    camera_x = camera_x[order]
    camera_intensity = camera_intensity[order]

    # Fit an offset added to the ion coordinates.
    # Require all ion points to remain inside camera coverage.
    lower = max(-max_offset_um, camera_x.min() - ion_x.min())
    upper = min(max_offset_um, camera_x.max() - ion_x.max())

    if lower >= upper:
        raise ValueError(
            "No usable offset range within camera coverage. "
            "Reduce the ion fitting range or increase max_offset_um."
        )

    def mse(offset):
        prediction = np.interp(
            ion_x + offset,
            camera_x,
            camera_intensity,
        )
        return np.mean((intensity - prediction) ** 2)

    # Coarse search followed by refinement around local minima.
    offsets = np.linspace(lower, upper, 81)
    errors = np.array([mse(offset) for offset in offsets])

    best_index = int(np.argmin(errors))
    best_offset = offsets[best_index]
    best_mse = errors[best_index]

    candidates = {best_index}
    candidates.update(
        k for k in range(1, len(offsets) - 1)
        if errors[k] <= errors[k - 1]
        and errors[k] <= errors[k + 1]
    )

    for k in candidates:
        fit = minimize_scalar(
            mse,
            bounds=(
                offsets[max(0, k - 1)],
                offsets[min(len(offsets) - 1, k + 1)],
            ),
            method="bounded",
        )

        if fit.success and fit.fun < best_mse:
            best_offset = fit.x
            best_mse = fit.fun

    return (
        ion_x + best_offset,
        camera_x,
        camera_intensity,
        best_offset,
        np.sqrt(best_mse),
    )


fig, axes = plt.subplots(
    1, 2,
    figsize=(12, 6),
    dpi=100,
    sharex=True,
    constrained_layout=True,
)

datasets = [
    (
        "Integrated $\lambda = 455$ nm delivery",
        axial_455_x_um,
        axial_455_intensity,
        axial_455_yerr,
        TE_450,
    ),
    (
        "Integrated $\lambda = 650$ nm delivery",
        axial_650_x_um,
        axial_650_intensity,
        axial_650_yerr,
        TE_635,
    ),
]

for ax, (title, ion_x, intensity, error, camera) in zip(axes, datasets):
    ion_x_aligned, camera_x, camera_y, offset, rmse = (
        align_ion_to_camera(
            ion_x,
            intensity,
            camera,
            max_offset_um=20,
        )
    )

    ax.errorbar(
        ion_x_aligned,
        intensity,
        yerr=error,
        fmt="o",
        mfc="none",
        color="black",
        capsize=3,
        label="Ion",
    )

    ax.plot(
        camera_x,
        camera_y,
        color="C0",
        linestyle="--",
        linewidth=2.5,
        label="Camera",
    )

    ax.set_title(title, fontsize=16)
    ax.legend(frameon=False, fontsize=14)

    print(title)
    print(f"  Ion offset after peak centering: {offset:+.2f} µm")
    print(f"  RMSE: {rmse:.4f}")

# Shared x-limits; adjust as needed.
axes[0].set_xlim(-15, 15)

fig.supxlabel(r"Axial position ($\mu$m)", fontsize=16)
fig.supylabel("Normalized intensity (arb. units)", fontsize=16)

for ax, label in zip(axes, ["a", "b"]):
    ax.text(
        -0.10, 1.04,
        label,
        transform=ax.transAxes,
        fontsize=16,
        fontweight="bold",
        ha="left",
        va="bottom",
        clip_on=False,
    )

fig.savefig(
    "/Users/yiyangzhi/Desktop/fig3.svg",
    dpi=300,
    bbox_inches="tight",
)
plt.show()
```
