## Data structure and coordinates

```text
R2C2/profile/                    # 405 and 880 nm
# R3C6/ uses the same structure for 450, 515, 635 nm; ion label chosen independently.
  ion_M1/                         # or ion_P1/
    405_nm/                       # also 450, 515, 635, 729, 880 when measured
      focused_<timestamp>_<id>.tiff
      focused_<timestamp>_<id>.json
      focused_<timestamp>_<id>_preview.png
      height_scan/
        NA0.55_Magnification50_000.tiff
        NA0.55_Magnification50_001.tiff
        ...
        exposure_log.csv
```

The log columns are `image_index`, `stage_position_mm`, `exposure_ms`, `max_intensity_8bit`, and `filename`. The JSON retains focus history, exposure, ROI, and `focus_state.result.final_z_mm`. The ROI describes autofocus selection; the observed TIFFs are full **1024 × 1280 × 3 RGB** images. The PNG is a preview; analysis reads the original TIFF.

`height_um = (stage_position_mm − focus_stage_mm) × 1000 × 0.5`

The first captured frame is about 1 µm above focus, and subsequent frames are about 1 µm apart after the stage ratio. Heights retain readback precision rather than being rounded. Each wavelength retains its own measured heights and frame count; the discovery inventory reports its current coverage.

| Object | Contents |
| --- | --- |
| `scans[405]` | `CaHeightScan` for the chosen ion; unmeasured slots such as `scans[635]` are `None` |
| `scans[405].frames` | DataFrame with original log columns plus TIFF `path` and calibrated `height_um` |
| `scans[405].reference_metadata` | Complete focused-image JSON |
| `white_light[405]` | Original reference pixels `[row, column, channel]`, no rotation |
| `sample_frames[405]` | First scan frame, original pixels and dtype |
| `intensity[405]` | Optional float32 stack `[frame, row, column]`, raw R+G+B counts |

The keys of `scans`, `white_light`, `sample_frames`, and `intensity` always include all six wavelengths. A `None` value indicates an unmeasured wavelength, or a stack that has not been loaded. Frame indices are acquisition labels, not heights. Rows increase downward and columns rightward; physical image axes are defined by the manual lens alignment below. `scan.device_name` identifies R2C2/R3C6/Gen1_OG; `scan.flip_vertical` is true only for R3C6 and is applied by the height-plane plotting loader.

## Read the original reference images and sample frames

`focused_*.tiff` is the candidate white-light reference for the next alignment step. The gallery allows visual inspection; no edge detection or rotation is applied yet. The full scans are read on demand to keep this first step lightweight.

````python
white_light = dict.fromkeys(WAVELENGTHS_NM)
sample_frames = dict.fromkeys(WAVELENGTHS_NM)
image_records = []
for wl, scan in scans.items():
    if scan is None:
        continue
    white_light[wl] = scan.read_reference()
    first_index = int(scan.frames.image_index.iloc[0])
    sample_frames[wl] = scan.read_frame(first_index)
    image_records.append({
        "wavelength_nm": wl,
        "device_name": scan.device_name, "ion_index": scan.ion_index,
        "flip_vertical_for_plot": scan.flip_vertical,
        "reference_shape": white_light[wl].shape,
        "scan_frame_shape": sample_frames[wl].shape,
        "dtype": str(sample_frames[wl].dtype),
        "first_image_index": first_index,
        "first_height_um": scan.frames.height_um.iloc[0],
    })
display(pd.DataFrame(image_records))
````

````python
fig, axes = plt.subplots(2, 3, figsize=(12, 7), constrained_layout=True)
for ax, wl in zip(axes.flat, WAVELENGTHS_NM):
    image = white_light[wl]
    scan = scans[wl]
    label = f"{scan.device_name} · ion {scan.ion_index}" if scan is not None else "Awaiting upload"
    ax.set_title(f"{label} · {wl} nm")
    if image is None:
        ax.text(0.5, 0.5, "Not measured yet", ha="center", va="center", transform=ax.transAxes)
        ax.set_axis_off()
    else:
        ax.imshow(image, origin="upper", cmap="gray")
        ax.set(xlabel="Column (pixels)", ylabel="Row (pixels)")
fig.suptitle("Focused references — original orientation")
plt.show()
````

## Choose the height and plot window

Choose a common `HEIGHT_UM` above the focused surface, or supply an entry in `HEIGHT_BY_WAVELENGTH_UM` to choose a different height for a wavelength. The nearest measured plane is selected **without interpolation**, within `MAX_HEIGHT_ERROR_UM`. The table reports the actual sampled height. Each panel title shows its wavelength; x/y axis labels are omitted, while numerical ticks and colorbars remain visible. A requested height beyond a scan's coverage raises a clear error instead of silently picking a distant plane. Use the discovery inventory to check the current height range for each device and wavelength.

The displayed window is centered on each beam and spans 20 µm in x (−10 to +10 µm) and 40 µm in y (−20 to +20 µm). `X_SPAN_UM` and `Y_SPAN_UM` control these full widths. Each panel is square, so one µm in x occupies twice the screen distance of one µm in y with these settings. The original ±30 µm normalization crop is retained, so zooming does not change the intensity values. Lens-relative coordinates are preserved in the result. To exclude another bright feature or a hot pixel, give `PEAK_ROI_UM[wl] = (xmin, xmax, ymin, ymax)` in aligned lens coordinates. To set the center directly, use `CENTER_UM[wl] = (x, y)` instead. The alignment previews show the lens origin in red. For R3C6, `PEAK_ROI_UM` and `CENTER_UM` use the vertically reflected lens coordinates (y changes sign relative to the unflipped preview). The mounting reflection affects image y only; it does not change the physical scan height.

Each panel is normalized to the maximum within its **±30 µm normalization crop**, giving a 0–1 scale before the narrower display window is applied. No automatic background subtraction or exposure correction is applied; the figure compares normalized shapes. Optional `BACKGROUND_COUNTS[wl]` subtracts a user-specified scalar in raw R+G+B units before clipping to zero. Exposure times, subtraction, chosen centers, and normalization peaks are recorded in the export.

Use `GLOBAL_X_SHIFT_UM` and `GLOBAL_Y_SHIFT_UM` for a common translation of every image panel, after each beam is centered. Positive X moves the beams right, positive Y moves them up; for example `(5, -3)` moves every beam 5 µm right and 3 µm down inside the display window. Change these values in the settings cell, then rerun the plot cell. Shifts are recorded in the export metadata and plotted x/y arrays.

````python
HEIGHT_UM =75.0  # Physical height above focus; change this and rerun the next cell.
HEIGHT_BY_WAVELENGTH_UM = {}  # e.g. {405: 70.0}; others use HEIGHT_UM.
MAX_HEIGHT_ERROR_UM = 0.6
HALF_WIDTH_UM = 30.0
X_SPAN_UM = 20.0  # Full displayed width: −10 to +10 µm.
Y_SPAN_UM = 40.0  # Full displayed height: −20 to +20 µm.
FIGURE_SIZE_INCHES = (11.5, 7.0)  # Three columns of square panels.
GLOBAL_X_SHIFT_UM = 0.0  # Positive moves every beam right; negative moves left.
GLOBAL_Y_SHIFT_UM = 0.0  # Positive moves every beam up; negative moves down.
SHOW_HEIGHTS = False  # Panel titles show wavelength without height.
FONT_SIZE = 18

PEAK_ROI_UM = dict.fromkeys(WAVELENGTHS_NM)  # Optional lens-relative search rectangle.
CENTER_UM = dict.fromkeys(WAVELENGTHS_NM)    # None centers on the brightest pixel; otherwise use lens-relative (x, y).
BACKGROUND_COUNTS = {wl: 0.0 for wl in WAVELENGTHS_NM}
````

````python
# Display with an opaque white canvas, including in a dark notebook theme.
%config InlineBackend.print_figure_kwargs = {"facecolor": "white", "edgecolor": "white", "transparent": False}

from importlib import reload
from matplotlib.colors import LinearSegmentedColormap
import Analysis.ca_3d_plotting as ca_plotting
plot_publication_grid = reload(ca_plotting).plot_publication_grid

# Recovered peak colors from paper_figures.ipynb. The 729 nm scan uses
# the paper's 730 nm color.
PAPER_PEAK_COLORS = {
    405: "#3700ff", 450: "#0413ff", 515: "#00ff90",
    635: "#ff000e", 729: "#ff4b17", 880: "#c87dff",
}

# Always read the selected source frame, avoiding a stale stack if ION_INDEX changed.
planes = dict.fromkeys(WAVELENGTHS_NM)
for wl, scan in scans.items():
    if scan is None:
        continue
    if alignments.get(wl) is None:
        raise RuntimeError(f"Complete manual alignment for {wl} nm first.")
    planes[wl] = select_height_plane(
        scan, alignments[wl], HEIGHT_BY_WAVELENGTH_UM.get(wl, HEIGHT_UM),
        max_height_error_um=MAX_HEIGHT_ERROR_UM,
        center_um=CENTER_UM[wl], peak_roi_um=PEAK_ROI_UM[wl],
        background_counts=BACKGROUND_COUNTS[wl],
    )

if "publication_figure" in globals():
    plt.close(publication_figure)
if (not np.isfinite([X_SPAN_UM, Y_SPAN_UM]).all()
        or min(X_SPAN_UM, Y_SPAN_UM) <= 0
        or max(X_SPAN_UM, Y_SPAN_UM) > 2 * HALF_WIDTH_UM):
    raise ValueError("Display spans must be positive and fit inside the normalization crop.")
publication_figure, publication_metadata, publication_arrays = plot_publication_grid(
    planes, half_width_um=HALF_WIDTH_UM, font_size=FONT_SIZE, show_heights=SHOW_HEIGHTS,
    x_shift_um=GLOBAL_X_SHIFT_UM, y_shift_um=GLOBAL_Y_SHIFT_UM,
)
publication_figure.set_size_inches(*FIGURE_SIZE_INCHES)
publication_figure.set_constrained_layout(False)
# The six panel axes are created before their colorbar axes. Update both
# image and attached colorbar through each image's mappable.
for ax, wl in zip(publication_figure.axes[:len(WAVELENGTHS_NM)], WAVELENGTHS_NM):
    ax.set(xlabel="", ylabel="",
           xlim=(-X_SPAN_UM/2, X_SPAN_UM/2), ylim=(-Y_SPAN_UM/2, Y_SPAN_UM/2))
    ax.set_title(f"λ = {wl} nm", fontfamily="Arial", fontsize=FONT_SIZE, pad=5)
    ax.set_aspect("auto")  # Different x/y screen scales preserve the requested limits.
    ax.set_box_aspect(1)    # Each image panel is square.
    ax.set_xticks(np.linspace(-X_SPAN_UM/2, X_SPAN_UM/2, 5))
    ax.set_yticks(np.linspace(-Y_SPAN_UM/2, Y_SPAN_UM/2, 5))
    for image in ax.images:
        image.set_cmap(LinearSegmentedColormap.from_list(
            f"paper_{wl}_nm", ["black", PAPER_PEAK_COLORS[wl]]))
# Position square panels and their colorbars together, leaving room for 18 pt ticks.
fig_w, fig_h = publication_figure.get_size_inches()
bar_width, bar_pad = 0.12, 0.10  # inches
column_gap, row_gap = 1.10, 0.75  # inches, including room for tick labels and titles
panel_size = min((fig_w - 1.10 - 3*(bar_width + bar_pad) - 2*column_gap)/3,
                 (fig_h - 1.00 - row_gap)/2)
if panel_size <= 0:
    raise ValueError("FIGURE_SIZE_INCHES is too small for this layout.")
grid_width = 3*(panel_size + bar_width + bar_pad) + 2*column_gap
left = (fig_w - grid_width)/2
bottom = (fig_h - (2*panel_size + row_gap))/2
for index, ax in enumerate(publication_figure.axes[:len(WAVELENGTHS_NM)]):
    row, col = divmod(index, 3)
    x = left + col*(panel_size + bar_pad + bar_width + column_gap)
    y = bottom + (1-row)*(panel_size + row_gap)
    ax.set_position((x/fig_w, y/fig_h, panel_size/fig_w, panel_size/fig_h))
    if ax.images:
        cbar = ax.images[0].colorbar
        cbar.ax.set_position(((x+panel_size+bar_pad)/fig_w, (y+0.05*panel_size)/fig_h,
                              bar_width/fig_w, 0.90*panel_size/fig_h))
for ax in publication_figure.axes:  # Include both panels and colorbars.
    ax.tick_params(axis="both", which="major", labelsize=FONT_SIZE)

# Shared position labels, centered on the six image panels. The tight export
# includes the extra label margins while preserving the square panel sizes.
publication_figure.supxlabel(
    "Axial position (µm)",
    x=(left + (grid_width - bar_pad - bar_width)/2)/fig_w,
    y=(bottom - 0.72)/fig_h,
    fontfamily="Arial", fontsize=FONT_SIZE,
)
publication_figure.supylabel(
    "Radial position (µm)",
    x=(left - 0.85)/fig_w,
    y=(bottom + panel_size + row_gap/2)/fig_h,
    fontfamily="Arial", fontsize=FONT_SIZE,
)
# Only the rightmost column needs the repeated intensity label.
for ax, wl in zip(publication_figure.axes[:len(WAVELENGTHS_NM)], WAVELENGTHS_NM):
    if wl in (515, 880) and ax.images:
        ax.images[0].colorbar.set_label(
            "Normalized intensity", fontfamily="Arial", fontsize=FONT_SIZE, labelpad=12,
        )

publication_metadata["peak_color_hex"] = publication_metadata["wavelength_nm"].map(PAPER_PEAK_COLORS)
publication_metadata["display_x_limits_um"] = [(-X_SPAN_UM/2, X_SPAN_UM/2)] * len(publication_metadata)
publication_metadata["display_y_limits_um"] = [(-Y_SPAN_UM/2, Y_SPAN_UM/2)] * len(publication_metadata)
publication_metadata["panel_box_aspect"] = 1.0  # Display height / width.

display(publication_metadata[[
    "device_name", "ion_index", "wavelength_nm", "flip_vertical",
    "requested_height_um", "selected_height_um",
    "frame_index", "exposure_ms", "center_um", "normalization_peak_counts", "peak_color_hex",
]])
display(publication_figure)
plt.close(publication_figure)
````

## Gaussian beam size versus wavelength

Run the height-plane plot above first. This analysis uses those same selected heights, manual rotations, R3C6 vertical reflections, and scalar background settings. For each available wavelength, `Experimental_Analysis.from_image` finds the **maximum pixel of the full aligned image** and extracts two single-pixel cuts through it: **axial = x (maximum's row)** and **radial = y (maximum's column)**. Display translations and the six-tile crop do not affect the fit.

Each cut is fit within `BEAM_FIT_HALF_WIDTH_UM` of its peak to **I(u) = B + A exp[−2(u−u₀)²/w²]**, including a fitted nonnegative constant background. The **1/e² intensity radius is w**. The table also reports **FWHM = √(2 ln 2) × w**. Set `BEAM_SIZE_QUANTITY = "fwhm"` to reproduce the width convention used by panel (b) in `paper_figures.ipynb`; `"radius"` plots beam radii. Colors distinguish wavelengths, with circles for radial and diamonds for axial.

Inspect the profile overlays and R² values before using a width. Failed fits have explicit status/error entries and no plotted point; `check fit window` means a fitted half-maximum falls outside the available fitting interval. Missing wavelengths remain absent. Fitted widths are point estimates; no averaging, smoothing, or pixel interpolation is added.

## Peak count rate versus height

For every measured frame, calculate

`peak_count_rate_per_s = max_over_pixels(R + G + B) / (exposure_ms / 1000)`

This is the brightest **single-pixel camera count rate**, in counts/s. Use the original full frame so rotation or cropping cannot discard a maximum; no manual alignment is needed for this calculation. A vertical flip preserves the maximum, so this calculation also uses the original R3C6 frames; peak row/column coordinates remain raw camera coordinates. There is no smoothing, background subtraction, or normalization to a curve's own maximum. Heights use each wavelength's focused zero and the 2:1 stage ratio, as above.

Each wavelength retains its own measured height range and vertical count-rate scale. These camera counts are exposure-normalized, but have not been corrected for wavelength-dependent camera response or input laser power. A circle marks the largest measured rate in each curve; the table lists that plane's actual height, frame index, exposure, and peak counts.

Run the calculation cell once after importing `scans`; it reads one TIFF at a time. Then rerun just the plot cell to change between `"linear"` and `"log"` display. The DataFrames in `peak_rate_tables` preserve all per-frame values and peak pixel coordinates.

````python
from importlib import reload
import Analysis.ca_3d_import as ca_import
import Analysis.ca_3d_plotting as ca_plotting
peak_count_rate_vs_height = reload(ca_import).peak_count_rate_vs_height
plot_peak_count_rates = reload(ca_plotting).plot_peak_count_rates

peak_rate_tables = dict.fromkeys(WAVELENGTHS_NM)
for wl, scan in scans.items():
    if scan is None:
        continue
    print(f"{scan.device_name}, ion {scan.ion_index}, {wl} nm: calculating peak rates for {len(scan.frames)} frames...", flush=True)
    peak_rate_tables[wl] = peak_count_rate_vs_height(scan)
peak_rate_data = pd.concat(
    [table for table in peak_rate_tables.values() if table is not None], ignore_index=True,
)
print(f"Calculated {len(peak_rate_data)} frame rates.")
````
