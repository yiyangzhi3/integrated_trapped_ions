# AlPad 639 etch on PECVD Standard fast oxide 1107

Mean thickness remaining as a function of total etch time.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

etch_data = pd.DataFrame({
    "Total time": [0, 0.5, 1.5, 3.5],
    "Mean thickness remaining": [3038, 2722, 2145, 1154],
})

x = etch_data["Total time"].to_numpy()
y = etch_data["Mean thickness remaining"].to_numpy()
slope, intercept = np.polyfit(x, y, 1)
y_pred = slope * x + intercept
r_squared = 1 - np.sum((y - y_pred) ** 2) / np.sum((y - y.mean()) ** 2)
x_fit = np.linspace(x.min(), x.max(), 200)

plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 18,
    "axes.labelsize": 18,
    "axes.titlesize": 20,
    "xtick.labelsize": 16,
    "ytick.labelsize": 16,
})

fig, ax = plt.subplots(figsize=(8.5, 6))
ax.scatter(x, y, s=90, color="#1F4E79", edgecolor="white", linewidth=1.2,
           zorder=3, label="Measured data (MSP300)")
ax.plot(x_fit, slope * x_fit + intercept, color="#D1495B", linewidth=2.5,
        label="Linear fit")

ax.set(
    title="ALPAD on PECVD SiO$_2$",
    xlabel="Total time (mins)",
    ylabel="Mean thickness remaining (nm)",
)
ax.text(
    0.04, 0.08,
    f"y = {slope:.1f}x + {intercept:.1f}\n$R^2$ = {r_squared:.4f}",
    transform=ax.transAxes,
    fontsize=16,
    bbox=dict(boxstyle="round,pad=0.35", facecolor="white", edgecolor="0.75", alpha=0.95),
)
ax.grid(axis="y", color="0.88", linewidth=0.8)
for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_linewidth(1.2)
ax.tick_params(direction="out", width=1.2, length=5)
ax.legend(frameon=False, fontsize=15)
ax.margins(x=0.04, y=0.08)
fig.tight_layout()
plt.show()

etch_data
```

```python
sheet_resistance = pd.DataFrame({
    "Process stage": ["After Al deposition", "After etch"],
    "Sheet resistance (Ω/square)": [0.075, 0.167],
})

fig, ax = plt.subplots(figsize=(8.5, 6))
bars = ax.bar(
    sheet_resistance["Process stage"],
    sheet_resistance["Sheet resistance (Ω/square)"],
    width=0.58,
    color=["#1F4E79", "#D1495B"],
    edgecolor="black",
    linewidth=1.2,
    zorder=3,
)

ax.bar_label(bars, labels=["0.075", "0.167"], padding=8, fontsize=17)
ax.set(
    title="Sheet Resistance after Processing",
    ylabel="Sheet resistance (Ω/square)",
    ylim=(0, 0.20),
)
ax.set_axisbelow(True)
ax.grid(axis="y", color="0.88", linewidth=0.8)
for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_linewidth(1.2)
ax.tick_params(direction="out", width=1.2, length=5)
fig.tight_layout()
plt.show()

sheet_resistance
```
